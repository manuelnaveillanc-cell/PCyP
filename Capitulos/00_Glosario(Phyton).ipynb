**Definición de función:**
> La instrucción (empezando con def) que usamos para crear una función nueva y guardarle un bloque de código reutilizable.

**Parámetro:**
> El nombre de la variable que ponemos entre paréntesis en la cabecera de la función para recibir los datos (argumentos) cuando la mandamos a llamar.

**Argumento:**
> El valor, dato o expresión que le pasas a una función cuando la mandas a llamar (por ejemplo, cuando le entregas un texto o un número entre paréntesis para que la función lo procese utilizando sus parámetros).

**Error de sintaxis (SyntaxError):**
> Pasa cuando rompes las reglas del código (como olvidarte de los dos puntos o un paréntesis). Python ni siquiera alcanza a ejecutarlo porque no entiende cómo leerlo.

**Error de nombre (NameError):**
> Ocurre cuando intentas usar una variable o función que no existe o que escribiste mal. Python te avisa que no reconoce ese nombre por ninguna parte.

**Error de tipo (TypeError):**
> Se da cuando mezclas datos incompatibles en una operación (como sumar un texto con un número), haciendo que Python no sepa cómo procesarlos.

**Error semántico:**
> Pasa cuando tu código corre perfecto y sin avisar errores técnicos, pero arroja un resultado distinto al que buscabas porque falló la lógica.

**Traceback:**
> El informe de error que te muestra Python cuando ocurre una falla en tiempo de ejecución, detallando paso a paso qué función llamó a cuál para entender dónde quedó la escoba.

**División entera (// vs /):**
> El operador / es la división normal y siempre devuelve un resultado decimal (float), aunque la división sea exactísima (por ejemplo, 84 / 2 da 42.0). En cambio, el operador // realiza una división entera redondeando siempre hacia abajo.

**Exponenciación (** **):
>El doble asterisco se usa para elevar un número a una potencia (por ejemplo, calcular el cuadrado escribiendo x ** 2), teniendo prioridad matemática por encima de las sumas y restas.

**Expresión vs Sentencia:**
> Una expresión (como n + 5) se evalúa y produce un valor como resultado. Una sentencia o instrucción (como n = 5) ejecuta una acción o efecto en el programa (como crear una variable), pero no necesariamente devuelve un valor visible por sí sola.

**Debugging:**
> El proceso metódico de buscar, analizar y corregir los errores o bugs que tenga tu código para que funcione como esperas.

**Marco (Frame):**
> Cada una de las cajitas dentro de un diagrama de pila que representa una llamada a función específica, conteniendo adentro sus variables locales y parámetros.

**Diagrama de pila (Stack diagram):**
> Un esquema gráfico que muestra las funciones que están activas en un momento dado, ordenadas en forma de pila.

**Dot operator (Operador punto .):**
> El punto que se usa para acceder a una función o variable que está guardada dentro de un módulo (por ejemplo, escribiendo math.sqrt).

**Módulo:**
> Un archivo externo que contiene funciones y herramientas ya hechas (como math o random). Para poder usarlas, primero debes cargarlas obligatoriamente usando la instrucción import.

**Keyword (Palabra reservada):**
> Palabras especiales que Python reserva exclusivamente para la estructura del lenguaje (como class, for, import o def). Nunca pueden usarse como nombres de variables, o de lo contrario se genera un error.

**Bucle (Loop):**
> Una estructura —como el ciclo for junto con range()— que sirve para repetir una o varias instrucciones de forma automática. Recuerda que en Python los rangos parten contando desde el 0.

**Variable local:**
> Una variable creada dentro de una función. Solo existe ahí dentro y se destruye apenas termina la ejecución; si intentas usarla afuera, Python te tira un error.
